```python
# prompt: 1. [5 marks] Create two vectors:
# v1 = [3, −2, 5], v2 = [1, 4, −1]
# Compute and display:
# (a) Dot product
# (b) Norm of each vector
# (c) Angle (in degrees) between them
# 2. [5 marks] Create a 3 × 3 matrix A and a 3 × 1 vector b:
# A =
# 
# 
# 2 −1 0
# 1 3 2
# 0 −2 1
# 
#  , b =
# 
# 
# 1
# 4
# −3
# 
# 
# Multiply A and b, and interpret the result.
# 3. [10 marks] Compute the following for matrix A:
# (a) Determinant
# (b) Rank
# (c) Inverse (if it exists)
# If the inverse exists, verify it by multiplying A and A−1

import numpy as np

# 1. Vector operations
v1 = np.array([3, -2, 5])
v2 = np.array([1, 4, -1])

# (a) Dot product
dot_product = np.dot(v1, v2)
print("(a) Dot product:", dot_product)

# (b) Norm of each vector
norm_v1 = np.linalg.norm(v1)
norm_v2 = np.linalg.norm(v2)
print("(b) Norm of v1:", norm_v1)
print("   Norm of v2:", norm_v2)

# (c) Angle between vectors
angle_rad = np.arccos(dot_product / (norm_v1 * norm_v2))
angle_deg = np.degrees(angle_rad)
print("(c) Angle between v1 and v2 (degrees):", angle_deg)


# 2. Matrix-vector multiplication
A = np.array([[2, -1, 0],
              [1, 3, 2],
              [0, -2, 1]])
b = np.array([[1],
              [4],
              [-3]])

result = np.dot(A, b)
print("\n2. Matrix-vector multiplication result:\n", result)
print("Interpretation: The result is a 3x1 vector representing a linear combination of the columns of A, with coefficients from vector b.")


# 3. Matrix operations
# (a) Determinant
determinant = np.linalg.det(A)
print("\n3. (a) Determinant of A:", determinant)

# (b) Rank
rank = np.linalg.matrix_rank(A)
print("(b) Rank of A:", rank)

# (c) Inverse (if it exists)
if determinant != 0:
    A_inv = np.linalg.inv(A)
    print("(c) Inverse of A:\n", A_inv)

    # Verification
    identity = np.dot(A, A_inv)
    print("Verification (A * A_inv):\n", identity)  # Should be close to the identity matrix
else:
    print("(c) Inverse of A does not exist (determinant is zero)")
```

```python
# prompt: 1. [10 marks] Define the matrix:
# C =
# 4 −2
# 1 1
# • Compute the eigenvalues and eigenvectors of C
# • Verify the result by checking Cv = λv for each eigenpair
# 2. [10 marks] If C is diagonalizable, find matrices P and D such that:
# C = P DP −1
# Verify your result using NumPy.

import numpy as np

# Define matrix C
C = np.array([[4, -2],
              [1, 1]])

# Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# Verify the result
for i in range(len(eigenvalues)):
    eigenvalue = eigenvalues[i]
    eigenvector = eigenvectors[:, i]
    result = np.dot(C, eigenvector)
    verification = eigenvalue * eigenvector
    print(f"\nVerification for eigenvalue {eigenvalue}:")
    print("Cv:", result)
    print("λv:", verification)
    print("Are they equal?", np.allclose(result, verification))


# 2. Diagonalization
# Check if the matrix is diagonalizable (eigenvectors must be linearly independent)
if len(eigenvalues) == np.linalg.matrix_rank(eigenvectors):
    print("\nMatrix C is diagonalizable.")
    # Form matrix P from eigenvectors and matrix D from eigenvalues
    P = eigenvectors
    D = np.diag(eigenvalues)

    # Verify C = PDP^(-1)
    C_reconstructed = np.dot(P, np.dot(D, np.linalg.inv(P)))
    print("\nReconstructed Matrix C:\n", C_reconstructed)
    print("\nOriginal Matrix C:\n", C)
    print("\nAre they equal?", np.allclose(C, C_reconstructed))
else:
    print("\nMatrix C is not diagonalizable.")
```

```python
# prompt: Create a 3D visualization (using matplotlib or another library) showing how a 3D vector
# transforms under a given 3 × 3 matrix (e.g., a rotation or scaling matrix).

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Example transformation matrix (rotation around z-axis by 45 degrees)
theta = np.pi / 4
rotation_matrix = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta), np.cos(theta), 0],
    [0, 0, 1]
])

# Original vector
original_vector = np.array([1, 0, 0])

# Transformed vector
transformed_vector = np.dot(rotation_matrix, original_vector)


# Create the 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the original vector
ax.quiver(0, 0, 0, original_vector[0], original_vector[1], original_vector[2], color='b', label='Original Vector')

# Plot the transformed vector
ax.quiver(0, 0, 0, transformed_vector[0], transformed_vector[1], transformed_vector[2], color='r', label='Transformed Vector')

# Set axis labels and limits
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_xlim([-1.5, 1.5])  # Adjust limits as needed
ax.set_ylim([-1.5, 1.5])
ax.set_zlim([-1.5, 1.5])
ax.legend()

# Display the plot
plt.title('3D Vector Transformation')
plt.show()
```
